```python
import pandas as pd

data = {
    'Customer_ID': [101, 102, 101, 103, 102],
    'Order_Date': ['2026-01-10', '2026-01-12', '2026-01-15', '2026-01-18', '2026-01-20'],
    'Product': ['Laptop', 'Mobile', 'Laptop', 'Tablet', 'Mobile'],
    'Quantity': [2, 1, 3, 2, 4]
}

order_data = pd.DataFrame(data)

# 1. Total orders by each customer
print("Total Orders by Customer:")
print(order_data.groupby('Customer_ID').size())

# 2. Average order quantity for each product
print("\nAverage Quantity by Product:")
print(order_data.groupby('Product')['Quantity'].mean())

# 3. Earliest and Latest Order Dates
order_data['Order_Date'] = pd.to_datetime(order_data['Order_Date'])

print("\nEarliest Order Date:", order_data['Order_Date'].min())
print("Latest Order Date:", order_data['Order_Date'].max())
```

```python
import pandas as pd

data = {
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Laptop',
                'Mobile', 'Tablet', 'Laptop'],
    'Quantity': [10, 20, 15, 25, 30, 10, 35],
    'Date': ['2026-10-05', '2026-10-10', '2026-10-12',
             '2026-10-15', '2026-10-18', '2026-10-20', '2026-10-25']
}

df = pd.DataFrame(data)

df['Date'] = pd.to_datetime(df['Date'])

# Find sales of the past month
last_month = df[df['Date'] >= '2026-10-01']

# Find top 5 products
top_products = last_month.groupby('Product')['Quantity'].sum()
top_products = top_products.sort_values(ascending=False).head(5)

print("Top 5 Best-Selling Products:")
print(top_products)
```

```python
import pandas as pd

data = {
    'Property_ID': [1, 2, 3, 4, 5],
    'Location': ['Chennai', 'Bangalore', 'Chennai', 'Hyderabad', 'Bangalore'],
    'Bedrooms': [3, 5, 2, 4, 6],
    'Area': [1200, 2000, 1000, 1500, 2500],
    'Price': [5000000, 8000000, 4000000, 6000000, 9000000]
}

property_data = pd.DataFrame(data)

# 1. Average price in each location
print("Average Price by Location:")
print(property_data.groupby('Location')['Price'].mean())

# 2. Properties with more than four bedrooms
print("\nProperties with more than 4 bedrooms:")
print(property_data[property_data['Bedrooms'] > 4])

# 3. Property with largest area
print("\nProperty with Largest Area:")
print(property_data.loc[property_data['Area'].idxmax()])
```

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [100, 150, 120, 180, 200, 250]

# Line Plot
plt.plot(months, sales, marker='o')
plt.title("Monthly Sales - Line Plot")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()

# Bar Plot
plt.bar(months, sales)
plt.title("Monthly Sales - Bar Plot")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

temperature = [25, 27, 30, 33, 35, 32, 30, 29, 28, 27, 26, 24]

rainfall = [10, 15, 20, 30, 50, 100, 150, 120, 80, 200, 180, 40]

# 1. Line Plot of Temperature
plt.plot(months, temperature, marker='o')
plt.title("Monthly Temperature")
plt.xlabel("Months")
plt.ylabel("Temperature (°C)")
plt.show()

# 2. Scatter Plot of Rainfall
plt.scatter(months, rainfall)
plt.title("Monthly Rainfall")
plt.xlabel("Months")
plt.ylabel("Rainfall (mm)")
plt.show()
```
