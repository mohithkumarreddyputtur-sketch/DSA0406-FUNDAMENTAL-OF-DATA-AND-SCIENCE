```python
import numpy as np

student_scores = np.array([
    [85, 90, 78, 88],
    [92, 85, 80, 75],
    [78, 88, 92, 85],
    [90, 95, 85, 80]
])

subjects = ["Math", "Science", "English", "History"]

avg = np.mean(student_scores, axis=0)

print("Average marks:", avg)

highest = np.argmax(avg)

print("Highest average subject:", subjects[highest])
```

```python
import numpy as np

sales = np.array([
    [100, 200, 300],
    [150, 250, 350],
    [120, 220, 320]
])

avg = np.mean(sales)

print("Average price:", avg)
```

```python
import numpy as np

house_data = np.array([
    [3, 1200, 200000],
    [5, 1800, 350000],
    [6, 2200, 450000],
    [4, 1500, 250000],
    [5, 2000, 400000]
])

result = house_data[house_data[:, 0] > 4]

avg = np.mean(result[:, 2])

print("Average sale price:", avg)
```

```python
import numpy as np

sales_data = np.array([10000, 15000, 18000, 20000])

total = np.sum(sales_data)

increase = ((sales_data[3] - sales_data[0]) / sales_data[0]) * 100

print("Total sales:", total)
print("Percentage increase:", increase, "%")
```

```python
import numpy as np

fuel_efficiency = np.array([25, 30, 35, 40, 45])

avg = np.mean(fuel_efficiency)

car1 = 30
car2 = 45

improvement = ((car2 - car1) / car1) * 100

print("Average fuel efficiency:", avg)
print("Percentage improvement:", improvement, "%")
```

```python
prices = [100, 200, 300]
quantities = [2, 1, 3]

discount = 10
tax = 5

total = 0

for i in range(len(prices)):
    total += prices[i] * quantities[i]

discount_amount = total * discount / 100

after_discount = total - discount_amount

tax_amount = after_discount * tax / 100

final_cost = after_discount + tax_amount

print("Total price:", total)
print("Discount:", discount_amount)
print("Tax:", tax_amount)
print("Final cost:", final_cost)
```
